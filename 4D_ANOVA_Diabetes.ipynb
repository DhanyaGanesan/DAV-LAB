```python
import pandas as pd
import numpy as np
from scipy.stats import ttest_1samp

# Load dataset
uci_diabetes = pd.read_csv("uci_diabetes.csv")

# Perform One-Sample T-Test for Glucose
t_stat, p_value = ttest_1samp(uci_diabetes["Glucose"], popmean=100)

print(f"T-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_value:.4f}")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import ttest_ind

# Load datasets
uci_diabetes = pd.read_csv("uci_diabetes.csv")
pima_diabetes = pd.read_csv("pima_diabetes.csv")

# Perform Independent T-Test

features = ["Glucose", "BloodPressure", "BMI"]

for feature in features:
    t_stat, p_value = ttest_ind(
        uci_diabetes[feature],
        pima_diabetes[feature],
        equal_var=False
    )

    print(f"{feature}")
    print(f"T-Statistic: {t_stat:.4f}")
    print(f"P-Value: {p_value:.4f}")
    print()
```

```python
# Interpret the T-Test results

alpha = 0.05

for feature in features:
    t_stat, p_value = ttest_ind(
        uci_diabetes[feature],
        pima_diabetes[feature],
        equal_var=False
    )

    print(f"{feature}:")

    if p_value < alpha:
        print("Significant difference between the two datasets.")
    else:
        print("No significant difference between the two datasets.")

    print()
```
