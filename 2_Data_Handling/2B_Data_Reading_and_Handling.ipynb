```python
import pandas as pd

# Create sample CSV data
csv_data = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [21, 22, None, 24, 23],
    "City": ["Chennai", "Bangalore", "Hyderabad", None, "Mumbai"]
})

csv_data.to_csv("Google_data (2b.c1).csv", index=False)

# Create sample Excel data
excel_data = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E"],
    "Marks": [90, 85, None, 70, 95],
    "Grade": ["A", "B", "A", None, "A"]
})

excel_data.to_excel("data (2c2).xlsx", sheet_name="Sheet1", index=False)

print("Sample CSV and Excel files created successfully!")
```

```python
import pandas as pd

# Read CSV file
text_df = pd.read_csv("Google_data (2b.c1).csv")

# Read Excel file
excel_df = pd.read_excel("data (2c2).xlsx", sheet_name="Sheet1")

# Read CSV from website
url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
web_df = pd.read_csv(url)

# Display first five rows
print("CSV Data")
print(text_df.head())

print("\nExcel Data")
print(excel_df.head())

print("\nWeb Data")
print(web_df.head())

# Handle missing values
text_df.ffill(inplace=True)
excel_df.bfill(inplace=True)
web_df.dropna(inplace=True)

# Save processed files
text_df.to_csv("processed_text.csv", index=False)
excel_df.to_excel("processed_excel.xlsx", index=False)

print("\nData processed successfully!")
```
